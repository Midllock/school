📓 Úkol: Manipulace s textem, kódováním a binárními daty
Tento notebook demonstruje převod textu na binární reprezentaci, základní šifrování pomocí operace XOR a konverzi kódování souborů.

1. Převod textu na binární řetězec a Hexadecimální zápis
Počítač neukládá písmena, ale čísla (bajty). To, jaké číslo patří kterému písmenu, určuje kódování (např. UTF-8). Pro přehlednost si binární data vypisujeme v šestnáctkové (hex) soustavě.

```python
# Definice textu se speciálními znaky
text = "Ahoj světe! 🚀"

# Převod na bajty (binární data) v různých kódováních
data_utf8 = text.encode('utf-8')
data_utf16 = text.encode('utf-16le')

print(f"Původní text: {text}")
print("-" * 40)
print(f"UTF-8 (hex výpis):    {data_utf8.hex(' ')}")
print(f"UTF-16LE (hex výpis): {data_utf16.hex(' ')}")
print("-" * 40)
print(f"Počet bajtů v UTF-8:  {len(data_utf8)}")
print(f"Počet bajtů v UTF-16: {len(data_utf16)}")
```

2. Šifrování a dešifrování pomocí bitového XOROperace XOR (Exclusive OR) je v kryptografii oblíbená, protože je reverzibilní. Pokud data "vy-xorujeme" stejným klíčem dvakrát, dostaneme původní hodnotu.Logika: $A \oplus K = B \implies B \oplus K = A$

```python
def xor_nastroj(data, klic):
    # Procházíme bajt po bajtu a aplikujeme XOR s klíčem (0-255)
    return bytes([bajt ^ klic for bajt in data])

# 1. Příprava dat
zprava = "Tajne Heslo 123".encode('utf-8')
moje_cislo_klice = 42  # Libovolné číslo od 0 do 255

# 2. Šifrování
sifra = xor_nastroj(zprava, moje_cislo_klice)
print(f"Zašifrovaná zpráva (hex): {sifra.hex(' ')}")

# 3. Dešifrování (použijeme úplně stejnou funkci i klíč)
puvodni_data = xor_nastroj(sifra, moje_cislo_klice)
print(f"Dešifrovaná zpráva:       {puvodni_data.decode('utf-8')}")
```

3. Změna kódování textového souboru
Při změně kódování (např. z UTF-8 na UTF-16) musíme soubor správně přečíst do paměti a následně jej uložit s novou instrukcí pro kódování.

```python
import os

# Nejdříve vytvoříme testovací soubor v UTF-8
with open('vstup_utf8.txt', 'w', encoding='utf-8') as f:
    f.write("Tento text měníme z UTF-8 na UTF-16 LE.")

# Funkce pro konverzi
def zkonvertuj_soubor(cesta, kodovani_z, kodovani_do):
    # Načtení v původním kódování
    with open(cesta, 'r', encoding=kodovani_z) as f:
        obsah = f.read()
    
    # Uložení v novém kódování
    nova_cesta = cesta.replace(kodovani_z, kodovani_do)
    with open(nova_cesta, 'w', encoding=kodovani_do) as f:
        f.write(obsah)
    
    return nova_cesta

# Provedení akce
vystup = zkonvertuj_soubor('vstup_utf8.txt', 'utf-8', 'utf-16le')

print(f"Soubor byl převeden a uložen jako: {vystup}")
print(f"Velikost původního: {os.path.getsize('vstup_utf8.txt')} bajtů")
print(f"Velikost nového:    {os.path.getsize(vystup)} bajtů")
```
